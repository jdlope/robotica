# Connect

Example of how to use the core classes to work with CoppeliaSim. The simulator must be ready to run a scene with the robot Pioneer 3-DX. The robot in the scene must use the label `PioneerP3DX`. The simulation is started and finished by the script. Note that API calls can still be made using the `coppelia.sim` property.

```python
from core import Coppelia
from core import P3DX

coppelia = Coppelia()
robot = P3DX(coppelia.sim, 'PioneerP3DX')

coppelia.start_simulation()
robot.set_speed(+1.2, -1.2)

while (t := coppelia.sim.getSimulationTime()) < 3:
    print(f'Simulation time: {t:.3f} [s]')

coppelia.stop_simulation()
```
