# Avoid

The script implements a simple rule-based heuristic controller for obstacle avoidance. The simulator must be ready to run a scene with the robot Pioneer 3-DX. The robot in the scene must use the label `PioneerP3DX`. The simulation is started and finished by the script.

## Import the classes

First, we need to import the classes from the module `core` to connect with the simulator and to operate the robot. The package `coppeliasim_core` must be installed (see the repository `README.md` for details).

```python
from core import Coppelia
from core import P3DX
```

## Function `avoid`

The function `avoid` returns the speeds of the left and right wheels to avoid collisions with obstacles and walls. It uses the `readings` provided by the sonar.

```python
def avoid(readings):
    if (readings[3] < 0.1) or (readings[4] < 0.2):
        lspeed, rspeed = +0.1, -0.8
    elif readings[1] < 0.1:
        lspeed, rspeed = +1.3, +0.6
    elif readings[5] < 0.4:
        lspeed, rspeed = +0.1, +0.9
    else:
        lspeed, rspeed = +1.5, +1.5
    return lspeed, rspeed
```

## Main loop

After establishing the connection with the simulator and the robot, the script starts the simulation and runs the main loop until the `STOP` button is pressed in the simulator. Then the script follows the typical _perception-planning-action_ loop.

```python
coppelia = Coppelia()
robot = P3DX(coppelia.sim, 'PioneerP3DX')

coppelia.start_simulation()

while coppelia.is_running():
    readings = robot.get_sonar()
    lspeed, rspeed = avoid(readings)
    robot.set_speed(lspeed, rspeed)

coppelia.stop_simulation()
```
